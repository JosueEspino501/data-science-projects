# Paso 2. Prepara los datos

```python
# 2.1 Reemplazar nombres de columnas a minúsculas
df.columns = df.columns.str.lower()

# 2.2 Convertir tipos de datos
df['year_of_release'] = pd.to_numeric(df['year_of_release'], errors='coerce')
df['na_sales'] = pd.to_numeric(df['na_sales'], errors='coerce')
df['eu_sales'] = pd.to_numeric(df['eu_sales'], errors='coerce')
df['jp_sales'] = pd.to_numeric(df['jp_sales'], errors='coerce')
df['other_sales'] = pd.to_numeric(df['other_sales'], errors='coerce')
df['critic_score'] = pd.to_numeric(df['critic_score'], errors='coerce')
df['user_score'] = pd.to_numeric(df['user_score'], errors='coerce')

# 2.3 Ventas totales
df['total_sales'] = (df['na_sales'] + 
                     df['eu_sales'] + 
                     df['jp_sales'] + 
                     df['other_sales'])

print("Tipos de datos después de la conversión:")
print(df.dtypes)

print("Columnas después de la preparación:")
print(df.columns.tolist())


print("\nPrimeras 5 filas con la nueva columna 'total_sales':")
display(df[['name', 'platform', 'year_of_release', 'na_sales', 'eu_sales', 
            'jp_sales', 'other_sales', 'total_sales']].head())

# Análisis de valores ausentes
missing_values = df.isnull().sum().sort_values(ascending=False)
missing_percentage = (df.isnull().sum() / len(df) * 100).round(2)

missing_table = pd.DataFrame({
    'Valores Ausentes': missing_values,
    'Porcentaje (%)': missing_percentage
})

print("Tabla de valores ausentes:")
display(missing_table)
```

# Paso 5: Prueba de hipótesis

```python
# Hipótesis 1: User_Score Xbox One == PC
xone = df_relevant[(df_relevant['platform'] == 'XOne') & df_relevant['user_score'].notna()]['user_score']
pc   = df_relevant[(df_relevant['platform'] == 'PC')   & df_relevant['user_score'].notna()]['user_score']

# Hipótesis nula H0: medias iguales
# Hipótesis alternativa H1: medias diferentes
alpha = 0.05

t_stat, p_value = stats.ttest_ind(xone, pc, equal_var=False)
print(f"Xbox One vs PC, p-value = {p_value:.5f}")
if p_value < alpha:
    print("Rechazamos H0, las medias son diferentes")
else:
    print("No rechazamos H0, las medias son iguales")
```

- Hipótesis nula (H0): El promedio de User_Score de Xbox One es igual al de PC.
- Hipótesis alternativa (H1): El promedio de User_Score de Xbox One es diferente al de PC.

```python
# Hipótesis 2: User_Score Action != Sports
action  = df_relevant[(df_relevant['genre'] == 'Action')  & df_relevant['user_score'].notna()]['user_score']
sports  = df_relevant[(df_relevant['genre'] == 'Sports')  & df_relevant['user_score'].notna()]['user_score']

t_stat2, p_value2 = stats.ttest_ind(action, sports, equal_var=False)
print(f"Action vs Sports, p-value = {p_value2:.5f}")
if p_value2 < alpha:
    print("Rechazamos H0, las medias son diferentes")
else:
    print("No rechazamos H0, las medias son iguales")
```

- Hipótesis nula (H0): El promedio de User_Score de Acción es igual al de Deportes.
- Hipótesis alternativa (H1): El promedio de User_Score de Acción es diferente al de Deportes.

<div class="alert alert-block alert-success">
<b>Comentario de Revisor            </b> <a class="tocSkip"></a>

Bien hecho con los tests, fueron realizados de forma correcta. 

Es importante mencionar que el t-test asume que las dos muestras tienen varianzas iguales Para verificar esa igualdad de varianzas podemos usar el **test de Levene**. Este nos dice si la variabilidad entre los grupos es *estadísticamente diferente*. En futuros notebooks podrías utilizar este test para decidir sobre la igualdad de varianzas. En corto, el flujo sería así

1. Usamos `levene()` para verificar si las varianzas son iguales.
2. Dependiendo del resultado, usamos `ttest_ind()` con el argumento `equal_var`:
   - `equal_var=True` si Levene dice que las varianzas son iguales (p > 0.05).
   - `equal_var=False` si Levene dice que las varianzas son diferentes (p < 0.05).

- **Hipótesis 1:** No se encontró diferencia significatva entre las calificaciones promedio de usuarios de **Xbox One** y **PC**. Ambas plataformas tienen puntuaciones similares.

#### Recomendaciones finales para la campaña 2017

4. **Adaptar** las campañas según la región (diferente mensaje y selección de juegos para Japón).
