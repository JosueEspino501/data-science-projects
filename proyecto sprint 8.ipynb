```python
# ── Paso 5
# Cargar archivo
df7 = pd.read_csv('/datasets/project_sql_result_07.csv')
# Inspección general
print(df7.info())
print(df7.head())


```

```python
# Separar grupos
rainy = df7[df7['weather_conditions'] == 'Bad']['duration_seconds']
good  = df7[df7['weather_conditions'] == 'Good']['duration_seconds']

# Prueba t de Student (dos muestras independientes)
alpha = 0.05
t_stat, p_value = stats.ttest_ind(rainy, good)

print(f'T-estadístico: {t_stat:.4f}')
print(f'P-valor: {p_value:.4f}')

if p_value < alpha:
    print('Rechazamos H0: la duración SÍ cambia en sábados lluviosos')
else:
    print('No rechazamos H0: no hay diferencia significativa')
```

## Prueba de Hipótesis

### Planteamiento de hipótesis
- **H0 (hipótesis nula):** La duración promedio de los viajes desde el Loop 
hasta el Aeropuerto O'Hare los sábados con mal clima es **igual** a la duración 
promedio en sábados con buen clima.
- **H1 (hipótesis alternativa):** La duración promedio de los viajes desde el Loop 
hasta el Aeropuerto O'Hare los sábados con mal clima es **diferente** a la duración 
promedio en sábados con buen clima.

### Criterio utilizado y por qué
Se utilizó la prueba t de Student para dos muestras independientes por las 
siguientes razones:
1. Los viajes en días lluviosos y en días buenos son grupos independientes entre sí.
2. Queremos comparar la duración promedio entre ambos grupos.
3. El nivel de significación establecido es **alfa = 0.05**.

### Criterio de decisión
- Si **p_value < 0.05** entonces rechazamos H0, el clima sí afecta la duración.
- Si **p_value >= 0.05** entonces no rechazamos H0, no hay evidencia suficiente.
